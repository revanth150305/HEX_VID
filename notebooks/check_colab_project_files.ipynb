# Check Course Project files in Google Drive

This notebook mounts Google Drive and checks the project code only. It deliberately does **not** inspect or upload the `data/` folder.

```python
from google.colab import drive
from pathlib import Path

drive.mount('/content/drive')

# Change this only if you stored the project in a different Drive folder.
PROJECT_DIR = Path('/content/drive/MyDrive/Course Project')

if not PROJECT_DIR.is_dir():
    raise FileNotFoundError(
        f'Project folder was not found: {PROJECT_DIR}\n'
        'In Google Drive, ensure it is in My Drive/Course Project, then run again.'
    )

print('Project folder found:', PROJECT_DIR)
print('Top-level entries (data excluded):')
for item in sorted(PROJECT_DIR.iterdir(), key=lambda path: path.name.lower()):
    if item.name == 'data':
        continue
    kind = 'folder' if item.is_dir() else 'file'
    print(f'  [{kind}] {item.name}')
```

```python
# These are the files/folders needed to set up and train in Colab.
# `data` is intentionally not listed or checked here.
required_paths = [
    'src',
    'scripts',
    'notebooks',
    'requirements.txt',
    'requirements-colab.txt',
    'scripts/setup_colab.sh',
    'scripts/verify_device.py',
    'scripts/train_supervised.py',
]

missing = []
for relative_path in required_paths:
    path = PROJECT_DIR / relative_path
    if path.exists():
        print(f'OK      {relative_path}')
    else:
        print(f'MISSING {relative_path}')
        missing.append(relative_path)

print()
if missing:
    print(f'Project code is incomplete: {len(missing)} required item(s) missing.')
else:
    print('Project code check passed. You can run the setup notebook/cells next.')
```

```python
# Optional: list every code/config file under the project, still excluding data and generated folders.
excluded_top_level = {'.venv', '.venv-directml', '__pycache__', 'data', 'checkpoints', 'results', 'logs'}
files = [
    path.relative_to(PROJECT_DIR)
    for path in PROJECT_DIR.rglob('*')
    if path.is_file() and not any(part in excluded_top_level for part in path.relative_to(PROJECT_DIR).parts)
]

print(f'Found {len(files)} code/config files (data excluded):')
for path in sorted(files, key=lambda value: str(value).lower()):
    print(' ', path)
```
